#### <b>Aufgabe (2): Extrahieren Sie die numerischen Werte der folgenden Variablen, und speichern Sie diese in separaten Variablen im DataFrame ab.</b>
<b>Details zur Aufgabenstellung:</b>
- Extrahieren Sie die numerischen Werte der folgenden Variablen, und speichern Sie diese in separaten Variablen ab:
  - price_raw
  - camera_resolution_raw
  - battery_life_raw
  - storage_size_raw
  - screen_size_raw<br>
- Benennen Sie die Spalten der neuen Variablen als (in Klammern ist der Datentyp angegeben):
  - price (int64)
  - camera_resolution (int64)
  - battery_life (int64)
  - storage_size (int64)
  - screen_size (float64)
- Stellen Sie ein Subset des DataFrames nur mit den neuen Spalten und den ersten 5 Zeilen im Jupyter Notebook dar.<br><br>
- Die folgenden regular expressions sind eine Hilfestellung zum Lösen der Aufgabe:<br>
  ```python
  "CHF\s*(\d+)\.-"
  "(\d+)\s*MP"
  "(\d+)\s*hrs"
  "(\d+)\s*GB"
  "(\d+\.\d+)\s*IN"
  ```
  
<b style="color: gray;">(max. erreichbare Punkte: 8)</b>

````python
# Extrahieren der numerischen Werte aus 'price_raw' und speichern in separaten Variablen
price = []
for i in df['price_raw']:                       #dieser Code durchläuft alle Einträge in der Spalte 'price_raw'
    d1 = re.findall("CHF\s*(\d+)\.-", i)        #--> diese Codefolge extrahiert Zahlen nach 'CHF' und vor '.-'
    try:
        d2= int(''.join(d1))                    #--> diese Codefolge fügt die extrahierten Zahlen zusammen
    except:
        d2 = None                               #dieser Code setzt den Wert auf None, wenn keine Zahl extrahiert werden kann
    price.append(d2)                            #dieser Code fügt den Wert (entweder Zahl oder None) zur Liste 'price' hinzu

# Save new variable in the pandas DataFrame
df['price'] = pd.Series(price, dtype='int64')   #--> diese Codefolge speichert die Liste 'price' als neue Spalte im DataFrame 'df' mit dem Datentyp 'int64'
````

````python
# Extrahieren der numerischen Werte aus 'camera_resolution_raw' und speichern in separaten Variablen
camera = []
for i in df['camera_resolution_raw']:                       #dieser Code durchläuft alle Einträge in der Spalte 'camera_resolution_raw'
    d1 = re.findall("(\d+)\s*MP", i)        #--> diese Codefolge extrahiert Zahlen vor 'MP'
    try:
        d2= int(''.join(d1))                    #--> diese Codefolge fügt die extrahierten Zahlen zusammen
    except:
        d2 = None                               #dieser Code setzt den Wert auf None, wenn keine Zahl extrahiert werden kann
    camera.append(d2)                          #dieser Code fügt den Wert (entweder Zahl oder None) zur Liste 'camera' hinzu

    # Save new variable in the pandas DataFrame
df['camera_resolution'] = pd.Series(camera, dtype='int64')   #--> diese Codefolge speichert die Liste 'price' als neue Spalte im DataFrame 'df' mit dem Datentyp 'int64'
````

````python
# Extrahieren der numerischen Werte aus 'battery_life_raw' und speichern in separaten Variablen
battery = []
for i in df['battery_life_raw']:                       #dieser Code durchläuft alle Einträge in der Spalte 'battery_life_raw'
    d1 = re.findall("(\d+)\s*hrs", i)                   #--> diese Codefolge extrahiert Zahlen vor 'hours'
    try:
        d2= int(''.join(d1))                            #--> diese Codefolge fügt die extrahierten Zahlen zusammen
    except:
        d2 = None                                        #dieser Code setzt den Wert auf None, wenn keine Zahl extrahiert werden kann
    battery.append(d2)                                  #dieser Code fügt den Wert (entweder Zahl oder None) zur Liste 'battery' hinzu

# Save new variable in the pandas DataFrame
df['battery_life'] = pd.Series(battery, dtype="int64")   #--> diese Codefolge speichert die Liste 'price' als neue Spalte im DataFrame 'df' mit dem Datentyp 'int64'
````

````python
# Extrahieren der numerischen Werte aus 'storage_size_raw' und speichern in separaten Variablen
storage =[]
for i in df['storage_size_raw']:                       #dieser Code durchläuft alle Einträge in der Spalte 'storage_size_raw'
    d1 = re.findall("(\d+)\s*GB", i)        #--> diese Codefolge extrahiert Zahlen vor 'GB'
    try:
        d2= int(''.join(d1))                    #--> diese Codefolge fügt die extrahierten Zahlen zusammen
    except:
        d2 = None                               #dieser Code setzt den Wert auf None, wenn keine Zahl extrahiert werden kann
    storage.append(d2)                          #dieser Code fügt den Wert (entweder Zahl oder None) zur Liste 'storage' hinzu

# Save new variable in the pandas DataFrame
df['storage_size'] = pd.Series(storage, dtype='int64')   #--> diese Codefolge speichert die Liste 'price' als neue Spalte im DataFrame 'df' mit dem Datentyp 'int64'
````

````python
# Extrahieren der numerischen Werte aus 'screen_size_raw' und speichern in separaten Variablen (Tipp: float64 statt int64)
screen = []
for i in df['screen_size_raw']:                       #dieser Code durchläuft alle Einträge in der Spalte 'screen_size_raw'
    d1 = re.findall("([\d\.]+)\s*IN", i)        #--> diese Codefolge extrahiert Zahlen (inkl. Dezimalstellen) vor 'inches'
    try:
        d2= float(''.join(d1))                    #--> diese Codefolge fügt die extrahierten Zahlen zusammen und wandelt sie in float um
    except:
        d2 = None                               #dieser Code setzt den Wert auf None, wenn keine Zahl extrahiert werden kann
    screen.append(d2)                          #dieser Code fügt den Wert (entweder Zahl oder None) zur Liste 'screen' hinzu    

# Save new variable in the pandas DataFrame
df['screen_size'] = pd.Series(screen, dtype='float64')   #--> diese Codefolge speichert die Liste 'price' als neue Spalte im DataFrame 'df' mit dem Datentyp 'float64'
````

#### <b>Aufgabe (7): Erstellen Sie ein Barchart mit der Anzahl Smartphones pro Hersteller.</b>
<b>Details zur Aufgabenstellung:</b>
- Ergänzen Sie den Python Code unten dort wo notwendig.
- Die Angabe zum Hersteller finden Sie in der Variable 'make'.
- Im Barchart müssen die Hersteller auf der vertikalen und die Anzahl Smartphones auf der horizontalen Achse abgebildet sein.
- Im Barchart müssen die Anzahl Smartphones pro Hersteller der Grösse nach absteigend sortiert sein.
- Beschriften Sie die Achsen des Barcharts und geben Sie dem Barchart einen Titel.<br><br>

<b style="color: gray;">(max. erreichbare Punkte: 8)</b>

````python
# Gruppierung der Daten nach 'make' und Berechnung der Anzahl Smartphones
df_bar = df['make'].value_counts().sort_values(ascending=True)          #mit dieser Codefolge werden die Daten im DataFrame 'df' nach der Spalte 'make' gruppiert, die Anzahl der Smartphones pro Hersteller gezählt und die Ergebnisse in aufsteigender Reihenfolge sortiert

# Zusammenstellen der Werte für den Barchart
napart = list(df_bar.values)            #napart enthält die Anzahl der Smartphones pro Hersteller
index = list(df_bar.index.values)       #index enthält die Namen der Hersteller
y_pos = np.arange(len(index))           #y_pos erstellt eine Liste von Positionen für die y-Achse basierend auf der Anzahl der Hersteller

# Erstellen des Barcharts
fig, ax = plt.subplots(figsize=(7,4))                             #Figsize definiert die Größe des Plots in Zoll / wobei 7 die Breite und 4 die Höhe des Plots
ax.barh(y_pos, napart, align = 'center', color = 'lightgreen')   #ax.barh erstellt einen horizontalen Balkendiagramm (Barchart) mit den Positionen y_pos und den Werten napart, die Balken sind zentriert und haben die Farbe 'lightgreen'
ax.set_yticks(y_pos, index)                                     #ax.set_yticks setzt die y-Achsen-Ticks auf die Positionen in y_pos und beschriftet sie mit den entsprechenden Herstellernamen aus index
ax.set_xlabel('Anzahl Smartphones', fontsize = 10)              #ax.set_xlabel fügt dem Barchart
ax.set_xlim(0, 1200)
ax.set_title('Anzahl Smartphones pro Hersteller', fontsize = 12)  #ax.set_title fügt dem Barchart einen Titel hinzu

# Anzeigen der Graphik
plt.show()
````
